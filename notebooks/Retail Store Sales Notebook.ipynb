```python
month_order= ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
plt.figure(figsize=(12, 6))
sns.barplot(data=data,x = 'Month',y = 'Sales',estimator=sum,order = month_order)
plt.title("Monthly Sales Overview",fontweight='bold')
plt.ylabel("Total Sales")
plt.show
```

```python
sns.barplot(data=data,x='Category',y= 'Sales',estimator=sum,color=' color='teal',)
plt.ylabel('Sales(M)')
plt.title('Total Sales by Catagory')
plt.show
```

```python
sns.barplot(data=data,x='Region',y= 'Sales',estimator=sum,color='seagreen',)
plt.ylabel('Sales(M)')
plt.title('Total Sales by Region')
plt.show
```

```python
sns.barplot(data=data,x='Segment',y= 'Sales',estimator=sum,color='darkorange',)
plt.ylabel('Sales(M)')
plt.title('Total Sales by Segment')
plt.show
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=data, x='Discount', y='Profit Margin %', hue='Category', alpha=0.6)
plt.title('Discount vs Profit Margin %')
plt.show()
```

```python
data['OrderMonth'] = data['Order Date'].dt.to_period('M').astype(str)
monthly = data.groupby('OrderMonth')[['Sales', 'Profit']].sum().reset_index()

fig, ax = plt.subplots(figsize=(16, 6))
ax.plot(monthly['OrderMonth'], monthly['Sales'], marker='o', label='Sales', color='steelblue')
ax.plot(monthly['OrderMonth'], monthly['Profit'], marker='o', label='Profit', color='darkorange')
ax.set_title('Monthly Sales & Profit Trend')
ax.set_xlabel('Month')
ax.set_ylabel('Amount (INR)')
ax.legend()
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=data,x='Sales',y='Product Name', color='mediumpurple',estimator='sum')
sns.set_style("darkgrid")
plt.title('Total Sales by Product')
plt.xlabel('Sales')
plt.show()
```

```python
sp_perf = data.groupby('Salesperson').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Orders=('Order ID', 'count')
).sort_values('Total_Sales', ascending=False)

sp_perf
```

```python
outlier_cols = ['Quantity', 'Unit Price', 'Discount', 'Sales', 'Profit', 'Profit Margin %']

fig, axes = plt.subplots(2, 3, figsize=(16, 8))
axes = axes.flatten()

for i, col in enumerate(outlier_cols):
    sns.boxplot(y=data[col], ax=axes[i], color='coral')
    axes[i].set_title(col)

plt.suptitle('Boxplots for Outlier Detection', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
return_by_cat = pd.crosstab(data['Category'], data['Return Status'], normalize='index') * 100

return_by_cat.plot(kind='bar', stacked=True, color=['seagreen', 'indianred'], figsize=(8, 5))
plt.title('Return Rate % by Category')
plt.ylabel('% of Orders')
plt.legend(title='Returned')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

- **Total Sales: $ 37,07,112 |Total Profit: $ 7,71,388 | Overall Margin: 20.81%**
- **Technology** dominates sales ( $ 27.7L), well ahead of Furniture ( $ 8.6L) and Office Supplies ( $ 0.75L)
- **South** is the top-selling region ($ 15.5L), followed by West ($ 14.0L) and North ($ 7.5L)
- **Consumer** is the leading segment ($ 13.8L), only slightly ahead of Home Office and Corporate - fairly balanced
- **Discount is negatively correlated with Profit Margin (r = -0.65)** - every markdown meaningfully erodes margin
- **43 orders (1.7%)** were sold at a loss (negative profit)
- **Return rate is ~50%** and evenly spread across categories - unusually high for real retail data, worth validating with the data source; may be a data-generation artifact rather than genuine customer behavior
- **Ship Mode doesn't track actual delivery speed** here - Express, Same Day, and Standard all average ~4 days to ship, so this field may not be reliable for delivery-time analysis
- **No missing values and no duplicate records** were found - the dataset is clean and ready for further modeling
